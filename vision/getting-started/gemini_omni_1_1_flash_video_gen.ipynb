## Overview

Gemini Omni 1.1 Flash is a model that understands the world around you, allowing you to animate photos or create videos from input media. Built on Gemini's world understanding and native multimodality, Gemini Omni 1.1 Flash creates outputs that reflect the logic of the real world and lets you shape them step-by-step through natural conversation.

In this tutorial, you'll learn how to use Gemini Omni 1.1 Flash in Gemini platform with the Google Gen AI SDK to try out the following scenarios:

- Video generation:
  - Text-to-video generation
  - Providing a reference image as a starting frame
  - Interpolating between starting and ending frames
  - Supplying reference images and videos to guide video generation
  - Stateless video generation
  - Streaming video generation
  - Async video generation
- Video editing:
  - Prompt-based video editing
  - Video extension
  - Multi-turn video editing (chat)
- Manage interactions:
  - Retrieve and delete an interaction
  - Cancel a background interaction

## Video generation

### Generate videos from a text prompt

With Gemini Omni 1.1 Flash, you can generate videos directly from text prompts via the [Interactions API](https://docs.cloud.google.com/gemini-enterprise-agent-platform/reference/models/interactions-api). To generate a video in the sample below, specify the following information:
- **Prompt:** A detailed description of the video you would like to see. For best results, consider attributes such as shot framing, motion, style, lighting, location, and action.
- **Generation Config:** Within this configuration, you can specify a `VideoConfig` that contains a `task` parameter.
  - **Task:** Valid options are `text_to_video`, `image_to_video`, `reference_to_video`, `extend`, or `edit`. Make sure to set each task accordingly based on your inputs and desired behavior.
- **Video Response Format:** Configure `aspect_ratio`, `duration`, `resolution`, and `delivery` parameters.
  - **Aspect ratio:** 16:9 or 9:16
  - **Duration:** 3s - 10s
  - **Resolution:** 360p, 720p, 1080p, 4k
  - **Delivery:** If you'd like to save your generated videos to GCS, set `delivery="uri"`. Additionally, set `gcs_uri="gs://<GCS_BUCKET>"`, making sure to replace `<GCS_BUCKET>` with your desired bucket path. To find your video's location, look inside the content list and copy the value next to `uri=` in the `VideoContent` object.


Notes on output videos:
- **Audio generation:** Audio will be generated alongside the output video.

All videos from `gemini-omni-1.1-flash-preview` include both [C2PA metadata](https://docs.cloud.google.com/gemini-enterprise-agent-platform/models/content-credentials) and a [SynthID watermark](https://deepmind.google/technologies/synthid/).

```python
prompt = "The words Gemini Omni Flash displayed boldly and centered. The words transform, cycling through different styles, fonts, and materials to align with the four seasons. The words remain perfectly positioned and constant while the colors, textures, and environments instantaneously change around it."

interaction = client.interactions.create(
    model=omni_model,
    input=prompt,
    generation_config=interactions.GenerationConfig(
        video_config=interactions.VideoConfig(
            task="text_to_video"
        )
    ),
    response_format=interactions.VideoResponseFormat(
        aspect_ratio="16:9",
        duration="6s",
        resolution="1080p",
        # delivery="uri",
        # gcs_uri="gs://<GCS_BUCKET>"
    )
)

contents = []
for step in interaction.steps:
  if step.type == "model_output":
    contents.extend(step.content)

show_video(contents[0].data)
```

#### Download the starting image

You can also generate a video by starting with an input image. In this example, you'll locally download an image that's stored in Google Cloud Storage. If you'd like, you can provide the URL of an image to display it below. If you have a local image you'd like to use, you can specify that in the following steps.

```python
!wget -q https://storage.googleapis.com/cloud-samples-data/generative-ai/image/suitcase.png
```

If you'd like to use a different local image, modify the file name in `start`.

```python
start = "suitcase.png"

fig, ax1 = plt.subplots(1, 1, figsize=(12, 6))
ax1.imshow(img.imread(start))
ax1.axis("off")
plt.show()
```

```python
prompt = "The suitcase stands vertically, unzips alongside and then opens in half. The wheels and shell stay solid. White bubbly, curvy retro text reading 'Omnicase' pops out of the center. Colorful travel stickers pop up around the text: a plane, palm tree, boat, city skyline, and a train."

with open(start, "rb") as f:
    img_b64 = base64.b64encode(f.read()).decode("utf-8")

interaction = client.interactions.create(
    model=omni_model,
    input=[
        {"type": "text", "text": prompt},
        {"type": "image", "mime_type": "image/png", "data": img_b64},
    ],
    generation_config=interactions.GenerationConfig(
        video_config=interactions.VideoConfig(
            task="image_to_video"
        )
    ),
    response_format=interactions.VideoResponseFormat(
        aspect_ratio="16:9",
        duration="6s",
        resolution="720p",
    )
)

contents = []
for step in interaction.steps:
  if step.type == "model_output":
    contents.extend(step.content)

show_video(contents[0].data)
```

### Interpolation

Interpolation generates a video that transitions smoothly between two images: a starting frame and an ending frame. You supply both images—the first becomes the opening frame of the video and the second becomes the closing frame—and the model synthesizes the in-between motion that connects them. The text prompt guides how the scene should move or how the camera should travel between the two frames.

Like generating from a single starting image, interpolation uses the `image_to_video` task. Providing a second image is what tells the model to interpolate toward that ending frame.

```python
!wget -q https://storage.googleapis.com/cloud-samples-data/generative-ai/image/river-1.png

!wget -q https://storage.googleapis.com/cloud-samples-data/generative-ai/image/river-2.png

start = "river-1.png"
end = "river-2.png"

# Display the images
fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(12, 6))
ax1.imshow(img.imread(start))
ax2.imshow(img.imread(end))
ax1.axis("off")
ax2.axis("off")
plt.show()
```

```python
prompt = "The camera moves around the bench to reveal the man's face. One continuous shot with no cuts."

with open(start, "rb") as f:
    start_img_b64 = base64.b64encode(f.read()).decode("utf-8")
with open(end, "rb") as f:
    end_img_b64 = base64.b64encode(f.read()).decode("utf-8")

interaction = client.interactions.create(
    model=omni_model,
    input=[
        {"type": "text", "text": prompt},
        {"type": "image", "mime_type": "image/png", "data": start_img_b64},
        {"type": "image", "mime_type": "image/png", "data": end_img_b64},
    ],
    generation_config=interactions.GenerationConfig(
        video_config=interactions.VideoConfig(
            task="image_to_video"
        )
    ),
    response_format=interactions.VideoResponseFormat(
        aspect_ratio="16:9",
        duration="9s",
        resolution="1080p",
    )
)

contents = []
for step in interaction.steps:
  if step.type == "model_output":
    contents.extend(step.content)

show_video(contents[0].data)
```

### Reference inputs in video generation

With Gemini Omni 1.1 Flash, you can use existing media to generate and edit new videos. The model processes these inputs in one of two ways:

  - **Source Media:** This content acts as the direct foundation for the final output.
    - *Examples:* In image-to-video, the provided image becomes the literal first frame. In video editing, the original video is modified via natural language prompts while keeping its core structure intact.
  - **Reference Media:** This content acts as a guide or inspiration rather than the literal base. You are asking the model to extract a specific style, subject, or mood to creatively reimagine it.
    - *Examples:* Storyboard frames, product images, style guides, audio jingles, or reference scenery. You can also use reference media for video editing—such as providing an image of a new character or art style to apply to a base video.


**NOTE:** Audio reference inputs are not currently supported.

#### Download and display reference images

In this example, you'll locally download one image and three videos that are stored in Google Cloud Storage.

```python
!wget -q https://storage.googleapis.com/cloud-samples-data/generative-ai/image/cartoon-robot.png

!wget -q https://storage.googleapis.com/cloud-samples-data/generative-ai/video/leapfrog.mp4

!wget -q https://storage.googleapis.com/cloud-samples-data/generative-ai/video/cartwheel.mp4

!wget -q https://storage.googleapis.com/cloud-samples-data/generative-ai/video/run.mp4
```

```python
character = "cartoon-robot.png"
movement1 = "leapfrog.mp4"
movement2 = "cartwheel.mp4"
movement3 = "run.mp4"

# Display the images
fig, ax1 = plt.subplots(1, 1, figsize=(8, 4))
ax1.imshow(img.imread(character))
ax1.axis("off")
plt.show()

display(Video("leapfrog.mp4", embed=True, width=400))
display(Video("cartwheel.mp4", embed=True, width=400))
display(Video("run.mp4", embed=True, width=400))
```

For best results, we recommend supplying no more than 5 reference images and 3 reference videos.

Additionally, you can specify aspect ratio, duration, or resolution through the text prompt. If including these details in the prompt, do not set these parameters in `response_format`.

```python
prompt = "A robot doing three movements back to back. 9:16 aspect ratio. 7 second video."

image_files = [character]
video_files = [movement1, movement2, movement3]

images_input = []
videos_input = []
for img_path in image_files:
    with open(img_path, "rb") as f:
        img_b64 = base64.b64encode(f.read()).decode("utf-8")
        images_input.append({"type": "image", "mime_type": "image/png", "data": img_b64})
for video_path in video_files:
    with open(video_path, "rb") as f:
        video_b64 = base64.b64encode(f.read()).decode("utf-8")
        videos_input.append({"type": "video", "mime_type": "video/mp4", "data": video_b64})

interaction = client.interactions.create(
    model=omni_model,
    input=[
        {"type": "text", "text": prompt},
        *images_input,
        *videos_input,
    ],
    generation_config=interactions.GenerationConfig(
        video_config=interactions.VideoConfig(
            task="reference_to_video"
        )
    ),
    response_format=interactions.VideoResponseFormat(
        resolution="720p",
    ),
)

contents = []
for step in interaction.steps:
  if step.type == "model_output":
    contents.extend(step.content)

show_video(contents[0].data)
```

### Stateless video generation

If you don't want an interaction to be persisted, set `store=False` when creating it. The interaction still runs and returns its output, but it isn't saved on the server, so you won't be able to retrieve it later or continue editing it in a follow-up turn.

```python
prompt = "A sleek laptop sits on a reflective white desk as clean 3D holographic bar charts and upward-trending line graphs rise into the air around it."

interaction = client.interactions.create(
    model=omni_model,
    input=prompt,
    store=False,
)

contents = []
for step in interaction.steps:
  if step.type == "model_output":
    contents.extend(step.content)

show_video(contents[0].data)
```

### Stream video generation

Rather than waiting for the complete response, you can stream an interaction to receive incremental updates while the model works. Set `stream=True` when creating the interaction and iterate over the resulting events. Each event carries a `delta`: you can surface the model's `thought_summary` as it reasons about your prompt, then display the `video` once generation is complete.

You can also set `store=True` to persist the interaction state, which lets you retrieve the result later or reference the interaction ID for multi-turn video editing.

```python
stream = client.interactions.create(
    model=omni_model,
    input="Create a video about a lonely robot who finds friendship in a most unexpected place.",
    response_format=[
        {"type": "video"}
    ],
    stream=True,
    store=True,
)

for event in stream:
    if hasattr(event, 'delta') and event.delta:
        if event.delta.type == 'thought_summary' and event.delta.content:
            display(Markdown(event.delta.content.text))
    if event.event_type == "step.delta":
        if event.delta.type == "video":
            show_video(event.delta.data)
    if event.event_type == "interaction.completed":
        interaction_id = event.interaction.id
        print("Interaction ID: " + interaction_id)
```

### Async video generation

To generate a video that can be checked on later, set the `background` parameter to True. You can then use `interactions.get` to check the status of the initial interaction ID.

```python
import time

prompt = "Raw cookie ingredients beautifully laid out on a kitchen counter. The view glides smoothly from one ingredient to the next. As each individual ingredient fills the view, a sleek 3D text graphic pops up to label it. The sequence finishes by resting intimately on the rich textures and details of the final ingredient after the label disappears."

initial_interaction = client.interactions.create(
    model=omni_model,
    input=prompt,
    background=True,
)
interaction = initial_interaction

while interaction.status not in ["completed", "failed"]:
    print(f"Current Status: {interaction.status}")
    time.sleep(10)
    interaction = client.interactions.get(id=initial_interaction.id)

if interaction.status == "completed":
    contents = []
    steps = getattr(interaction, "steps", []) or []
    for step in steps:
        if step.type == "model_output":
            contents.extend(step.content)
    show_video(contents[0].data)
else:
    print(f"Status: {interaction.status}")
```

### Extension

Video extension continues an existing video: the model takes your input video and generates new footage that picks up where it leaves off, keeping the characters, setting, and style consistent. The text prompt describes what should happen next in the continuation.

To extend a video, use the `extend` task with a source video (up to 30s long). Each call can add a maximum of 10s of new footage.

```python
prompt = "The dog runs in the office and barks at the front desk to get a treat."

with open(video_file, "rb") as f:
    video_b64 = base64.b64encode(f.read()).decode("utf-8")

interaction = client.interactions.create(
    model=omni_model,
    input=[
        {"type": "text", "text": prompt},
        {"type": "video", "mime_type": "video/mp4", "data": video_b64},
    ],
    generation_config=interactions.GenerationConfig(
        video_config=interactions.VideoConfig(
            task="extend"
        )
    ),
    response_format=interactions.VideoResponseFormat(
        duration="10s",
    ),
)

contents = []
for step in interaction.steps:
  if step.type == "model_output":
    contents.extend(step.content)

show_video(contents[0].data)
```

You can iteratively chat with the model to continuously change aspects of the video using the example below.

## Manage interactions

By default, the Interactions API stores each interaction so you can retrieve, edit, or delete it later. The following examples show how to manage the lifecycle of an interaction: deleting it, retrieving it by ID, and cancelling a running background interaction.

### Retrieve and delete an interaction

Every interaction you create is assigned a unique ID that you can use to retrieve it later with `client.interactions.get()`. When you no longer need an interaction, remove it from the server with `client.interactions.delete()`. After an interaction is deleted, attempting to retrieve it by that ID returns an error.

```python
prompt = " A fluffy ginger kitten carefully climbing the gnarled trunk of an old oak tree, tiny claws digging into the bark as it pulls itself higher in a single continuous handheld shot."

interaction = client.interactions.create(
    model=omni_model,
    input=prompt,
)

contents = []
for step in interaction.steps:
  if step.type == "model_output":
    contents.extend(step.content)

show_video(contents[0].data)
```

```python
client.interactions.delete(id=interaction.id)
```

Now that the interaction has been deleted, try to retrieve it again by its ID. Because the interaction no longer exists on the server, running the following cell is expected to raise an error.

```python
from rich.console import Console
console = Console()

try:
    interaction = client.interactions.get(id=interaction.id)
except Exception as e:
    console.print(f"{e}")
```

### Cancel a background interaction

For long-running background interactions, you can stop execution before it finishes by calling `client.interactions.cancel()`. The interaction's status then transitions to `cancelled`.

```python
import time

prompt = "A sweeping cinematic drone shot flying low over a misty mountain valley at sunrise, gliding above a winding river and pine forest as golden light breaks over the distant peaks. Wisps of fog drift through the trees. Sound design: gentle wind, distant birdsong, and a soft orchestral swell. No dialogue."

initial_interaction = client.interactions.create(
    model=omni_model,
    input=prompt,
    background=True,
)

client.interactions.cancel(id=initial_interaction.id)

interaction = client.interactions.get(id=initial_interaction.id)
while interaction.status not in ["completed", "failed", "cancelled"]:
    print(f"Current Status: {interaction.status}")
    time.sleep(5)
    interaction = client.interactions.get(id=initial_interaction.id)

print(f"Final Status: {interaction.status}")
```
